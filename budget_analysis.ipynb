# Feature eng 

```python
df = df.sort_values("date").reset_index(drop=True)
```

```python
df["date"].min(), df["date"].max() # start and end 
```

```python
monthly = df[df["transaction_type"] == "expense"].groupby(df["date"].dt.to_period("M"))["amount"].sum().reset_index() # date into year and month 
```

```python
monthly.columns = ["month", "monthly_expense"]
```

```python
monthly.head()
```

```python
for i in range(1, 7):
    monthly[f"lag_{i}"] = monthly["monthly_expense"].shift(i)

monthly["target"] = monthly["monthly_expense"].shift(-1)
```

```python
monthly.head(10)
```

```python
monthly["rolling_mean_3"] = (
    monthly["monthly_expense"].shift(1).rolling(3).mean()
)

monthly["rolling_mean_6"] = (
    monthly["monthly_expense"].shift(1).rolling(6).mean()
)
```

```python
monthly["rolling_std_3"] = (
    monthly["monthly_expense"].shift(1).rolling(3).std()
)

monthly["rolling_std_6"] = (
    monthly["monthly_expense"].shift(1).rolling(6).std()
)
```

```python
monthly["month_num"] = monthly["month"].dt.month
monthly["quarter"] = monthly["month"].dt.quarter
monthly["year"] = monthly["month"].dt.year
```

```python
monthly = monthly.dropna().reset_index(drop=True)
```

```python
monthly.head()
```

```python
X = monthly[
    [
        "lag_1", "lag_2", "lag_3",
        "lag_4", "lag_5", "lag_6",
        "rolling_mean_3", "rolling_mean_6",
        "rolling_std_3", "rolling_std_6",
        "month_num", "quarter", "year"
    ] 
]

y = monthly["target"]
```

```python
X.head()
```

```python
split = int(len(X) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=500,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred) ** 0.5
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
pd.DataFrame({
    "actual": y_test.values,
    "predicted": y_pred
})
```

## Visualization

```python
#Monthly expenses

plt.figure(figsize=(12, 5))

plt.plot(monthly["month"].astype(str), monthly["monthly_expense"])

plt.title("Monthly Expense Trend")
plt.xlabel("Month")
plt.ylabel("Total Expense (₹)")

#every 3rd month on the x-axis
plt.xticks(range(0, len(monthly), 3),
           monthly["month"].astype(str).iloc[::3],
           rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Category-wise Expense 

category_expense =(
    df[df["transaction_type"]== "expense"]
    .groupby("category")["amount"]
    .sum()
    .sort_values(ascending=False)
)

category_expense
```

```python
#Rent is largest 
plt.figure(figsize=(10, 5))

category_expense.plot(kind="bar")

plt.title("Expense by Category")
plt.xlabel("Category")
plt.ylabel("Total Expense (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
payment_expense = (
    df[df["transaction_type"] == "expense"]
    .groupby("payment_mode")["amount"]
    .sum()
    .sort_values(ascending=False)
)

payment_expense
```

```python
plt.figure(figsize=(8, 5))

payment_expense.plot(kind="bar")

plt.title("Expense by Payment Mode")
plt.xlabel("Payment Mode")
plt.ylabel("Total Expense (₹)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
monthly_income_expense = (
    df.groupby([df["date"].dt.to_period("M"), "transaction_type"])["amount"]
    .sum()
    .unstack(fill_value=0)
)

monthly_income_expense
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_income_expense.index.astype(str),
    monthly_income_expense["income"],
    label="Income"
)

plt.plot(
    monthly_income_expense.index.astype(str),
    monthly_income_expense["expense"],
    label="Expense"
)

plt.title("Monthly Income vs Expense")
plt.xlabel("Month")
plt.ylabel("Amount (₹)")
plt.legend()

plt.xticks(
    range(0, len(monthly_income_expense), 3),
    monthly_income_expense.index.astype(str)[::3],
    rotation=45
)

plt.tight_layout()
plt.show()
```
